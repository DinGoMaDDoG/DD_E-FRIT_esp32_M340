# Normalisation ranges for the M340 query vector

Prints the reciprocal data ranges (1/range) of setpoint, speed and controller output in the initial M340 run. These constants normalise the KNN distance.

```python
import numpy as np
import pandas as pd

df  = pd.read_csv("../data/m340/initial_run.csv")
data = df.values

sp_range =data[:, 0].max() - data[:, 0].min()
rpm_range=data[:, 1].max() - data[:, 1].min()
out_range= data[:, 2].max() - data[:, 2].min()
print(f"1/{sp_range:.2f}, 1/{rpm_range:.2f}, 1/{out_range:.2f}")

print(data[:, 1].max())
```
