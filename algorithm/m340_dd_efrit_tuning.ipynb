# DD-E-FRIT offline tuning: Modicon M340 + ATV630 drive

Same method as `esp32_dd_efrit_tuning.ipynb`, applied to data logged from the M340 PLC over Modbus TCP with `m340/tools/modbus_logger.py`.

Differences from the ESP32 version:
- The fictitious reference is derived for the PLC's I-PD law, which scales the integral gain by Ts and the derivative gain by 1/Ts.
- Reference-model and learning parameters were re-selected for the drive; gradients are clipped and the gain limits are much tighter.
- The query features are normalised offline (feature matrix), so the PLC only computes distances at run time.

> Generated files are written to `output/m340/`. The outputs shown are from the original run; the notebook was not re-executed after the repository was reorganised.

## 2. E-FRIT tuning of the database

Query vector φ(t) = [r(t+1), r(t), y(t), y(t−1), u(t−1), u(t−2)], k = 20 nearest neighbours, gradient descent on the E-FRIT cost (tracking error to the reference model plus a λ-weighted control-effort term).

```python
# ---- Tuning parameters -------------------------------------------------------
# ny, nu: number of past outputs / inputs in the query vector (documentation only;
#         the vector itself is built explicitly in buildVector below)
ny=2
nu=3
k=20                  # number of nearest neighbours
ts=0.3                # sample time [s]
risetime=0.8          # reference model: desired rise time [s]
damping=0             # reference model: damping (0..1)
max_epochs=1000       # gradient-descent iterations per record (upper limit)
lr_kp=0.000002        # learning rates
lr_ki=0.000002
lr_kd=0.00000002
l=0.0005              # lambda: weight of the control-effort term

# Discrete second-order reference model Td(z) = P_1*z^-1 / (1 + p1*z^-1 + p2*z^-2),
# shaped by the desired rise time and damping.
def calculateReferenceModel (ts, risetime, damping):
    rho=ts/risetime
    mu=0.25*(1-damping)+0.51*damping
    p1=-2*np.exp(-rho/(2*mu))*np.cos(np.sqrt(4*mu-1)*rho/(2*mu))
    p2=np.exp(-rho/mu)
    P_1=1+p1+p2

    return p1, p2, P_1 

# Differences of the closed-loop data at sample i: dy, du and d2y.
def checkDifferentials(y0, u0, i):
    dy0=y0[i]-y0[i-1]
    du0=u0[i]-u0[i-1]

    d2y=y0[i]-2*y0[i-1]+y0[i-2]

    return dy0, du0, d2y

# Query vector phi(i) = [r(i+1), r(i), y(i), y(i-1), u(i-1), u(i-2)].
def buildVector(i):
    return [setpoint[i+1], setpoint[i], rpm[i], rpm[i-1], pwm[i-1], pwm[i-2]]

# L1 distance between two query vectors, each element scaled by its data range.
def distanceCalc(queryPhi, infoPhi):
    total=0

    for q, d, r in zip(queryPhi, infoPhi, ranges):
        total+=abs(q-d)/r

    return total

# Closed-loop data (r, y, u) and the gains that were active when it was recorded.
setpoint=df["SetPoint"].values
rpm=df["RPM"].values
pwm=df["PWM"].values
kp=df["Kp"].values
ki=df["Ki"].values
kd=df["Kd"].values

# Normalisation ranges for the six elements of the query vector.
ranges=[
    setpoint.max()-setpoint.min(),
    setpoint.max()-setpoint.min(),
    rpm.max()-rpm.min(),
    rpm.max()-rpm.min(),
    pwm.max()-pwm.min(),
    pwm.max()-pwm.min(),
]

# The tuned database starts as a copy of the recorded gains.
optimized_kp=np.copy(kp)
optimized_ki=np.copy(ki)
optimized_kd=np.copy(kd)

p1, p2, P_1=calculateReferenceModel(ts, risetime, damping)

# fs scales the input-increment term to the same magnitude as the tracking term.
fs=np.sqrt(np.var(setpoint-rpm)/np.var(np.diff(pwm)))

# ---- E-FRIT tuning of every database record ---------------------------------
for query_idx in range (2, len(setpoint)-1):
    # 1) k nearest neighbours of this record (excluding itself)
    distances=[]

    infoVector=buildVector(query_idx)
    
    for j in range(2, len(setpoint)-1):
        if query_idx==j:
            continue
        
        query=buildVector(j)

        d=distanceCalc(query, infoVector)
        distances.append((d,j))

    distances.sort(key=lambda x: x[0])
    neighbors=distances[:k]

    # 2) Initial gains: weighted average of the neighbours' gains, weights ~ exp(-distance)
    exp_values=[np.exp(-n[0]) for n in neighbors]
    total_exp=sum(exp_values)
    weights=[e/total_exp for e in exp_values]

    kp_opt=sum(weights[w]*kp[neighbors[w][1]] for w in range(k))
    ki_opt=sum(weights[w]*ki[neighbors[w][1]] for w in range(k))
    kd_opt=sum(weights[w]*kd[neighbors[w][1]] for w in range(k))

    print(f"Optimizing Row {query_idx}")
    print(f"Initial Local PID: {kp_opt:.4f}, {ki_opt:.4f}, {kd_opt:.4f}")

    r_tilda=np.zeros(len(setpoint))
    y_ref=np.zeros(len(setpoint))
    x0=np.zeros(len(setpoint))
    diff_u_tilda=np.zeros(len(setpoint))

    prev_epoch=np.inf

    # 3) Gradient descent on the weighted E-FRIT cost
    for epoch in range(max_epochs):
        epoch_cost=0
        kp_grad=0
        ki_grad=0
        kd_grad=0
    

        # Fictitious reference r~ for the current gains and its reference-model response y_ref
        for i in range(2, len(setpoint)):
            dy0, du0, d2y=checkDifferentials(rpm, pwm, i)

            # ESP32 form (no sample-time scaling): r_tilda[i]=rpm[i]+(du0+kp_opt*dy0+kd_opt*d2y)/ki_opt
            # M340 form: integral gain scaled by Ts, derivative gain by 1/Ts
            r_tilda[i] = rpm[i] + (
                du0
                + kp_opt * dy0
                + (kd_opt / ts) * d2y
            ) / (ki_opt * ts)
            y_ref[i]=r_tilda[i-1]*P_1-y_ref[i-1]*p1-y_ref[i-2]*p2

    
        for w_idx in range(k):
            idx=neighbors[w_idx][1]
            w=weights[w_idx]

            if idx >= len(setpoint)-1: 
                continue

            dy0, du0, d2y=checkDifferentials(rpm, pwm, idx)
            dy_ref=y_ref[idx]-y_ref[idx-1]
            d2y_ref=y_ref[idx]-2*y_ref[idx-1]+y_ref[idx-2]

            # ESP32 form: x0[idx]=du0+kp_opt*dy0+kd_opt*d2y
            x0[idx] = (
                du0
                + kp_opt * dy0
                + (kd_opt / ts) * d2y
            )
            # ESP32 form: diff_u_tilda[idx]=ki_opt*(r_tilda[idx]-y_ref[idx])-kp_opt*dy_ref-kd_opt*d2y_ref
            diff_u_tilda[idx] = (
                ki_opt * ts * (r_tilda[idx] - y_ref[idx])
                - kp_opt * dy_ref
                - (kd_opt / ts) * d2y_ref
            )

            err=rpm[idx+1]-y_ref[idx+1] 

            # Cost: tracking error + (l*fs)-weighted fictitious input increment
            epoch_cost+=w*0.5*((err)**2+l*fs*(diff_u_tilda[idx])**2) 

            kp_grad+=w*(-err*P_1*dy0/ki_opt+l*fs*diff_u_tilda[idx]*(dy0-dy_ref))
            ki_grad+=w*(err*P_1*x0[idx]/(ki_opt**2)+l*fs*diff_u_tilda[idx]*(rpm[idx]-y_ref[idx]))
            kd_grad+=w*(-err*P_1*d2y/ki_opt+l*fs*diff_u_tilda[idx]*(d2y-d2y_ref))
        
        # Clip the gradients to keep the updates stable (earlier, looser limits:)
        # kp_grad = np.clip(kp_grad, -100.0, 100.0)
        # ki_grad = np.clip(ki_grad, -100.0, 100.0)
        # kd_grad = np.clip(kd_grad, -20.0, 20.0)

        kp_grad = np.clip(kp_grad, -20.0, 20.0)
        ki_grad = np.clip(ki_grad, -20.0, 20.0)
        kd_grad = np.clip(kd_grad, -5.0, 5.0)

        # Gradient step, then clamp the gains to a safe range
        kp_opt=kp_opt-(lr_kp*kp_grad)
        ki_opt=ki_opt-(lr_ki*ki_grad)
        kd_opt=kd_opt-(lr_kd*kd_grad)

        if kp_opt < 0.005:
            kp_opt = 0.005
        elif kp_opt > 0.030:
            kp_opt = 0.030

        if ki_opt < 0.005:
            ki_opt = 0.005
        elif ki_opt > 0.050:
            ki_opt = 0.050

        if kd_opt < 0.0000:
            kd_opt = 0.0000
        elif kd_opt > 0.0010:
            kd_opt = 0.0010

        if epoch % 25 == 0:
            print(f"Epoch {epoch}| Cost:{epoch_cost:.6f}| PID: {kp_opt:.6f}, {ki_opt:.6f}, {kd_opt:.6f}")

        # Alternative stopping rule (relative change), not used:
        # if epoch > 30:
        #     rel_change = np.abs(prev_epoch - epoch_cost) / (np.abs(prev_epoch) + 1e-9)

        #     if rel_change < 1e-5:
        #         print(f'Converged at: {epoch}')
        #         break

        # Stop once the cost changes by less than 0.01 between epochs
        if np.abs(prev_epoch-epoch_cost) < 0.01:
            print(f'Converged at: {epoch}')
            break

        prev_epoch = epoch_cost
    
    optimized_kp[query_idx]=kp_opt
    optimized_ki[query_idx]=ki_opt
    optimized_kd[query_idx]=kd_opt
    print(f"Row {query_idx} Final PID: {kp_opt:.4f}, {ki_opt:.4f}, {kd_opt:.4f}")

# Rows 0-1 and the last row have no complete query vector, so they are dropped.
new_dataset={
    "SetPoint": setpoint,
    "RPM": rpm,
    "PWM": pwm,
    "Kp": optimized_kp,
    "Ki": optimized_ki,
    "Kd": optimized_kd
}

optimized_df = pd.DataFrame(new_dataset)
optimized_df = optimized_df.iloc[2:-1].reset_index(drop=True)
optimized_df.to_csv("output/m340/DD_E_FRIT_Optimized_Dataset.csv", index=False)
```

## 3. Export for the PLC

### Structured Text: tuned database

```python
# Write the tuned database (raw values) as Structured Text initialisation code.
def make_st():
    df=pd.read_csv("output/m340/DD_E_FRIT_Optimized_Dataset.csv")
    dataset_size=len(df)

    with open("output/m340/data_init.txt", "w", newline="\n") as st_file:
        st_file.write ("IF %S21 THEN \n")
        st_file.write(f"    DATASET_SIZE := {dataset_size};\n")
        st_file.write("    (*Structure: {SetPoint, RPM, PWM} *)\n\n")

        for i, row in df.iterrows():
            st_file.write(f"    infoMatrix[{i},0] := {row['SetPoint']:.4f};\n")
            st_file.write(f"    infoMatrix[{i},1] := {row['RPM']:.4f};\n")
            st_file.write(f"    infoMatrix[{i},2] := {row['PWM']:.4f};\n")
            st_file.write(f"    pidMatrix[{i},0] := {row['Kp']:.6f};\n")
            st_file.write(f"    pidMatrix[{i},1] := {row['Ki']:.6f};\n")
            st_file.write(f"    pidMatrix[{i},2] := {row['Kd']:.6f};\n")
            st_file.write("     \n")
        
        st_file.write(f"    bDataLoaded := TRUE;\n")
        st_file.write ("END_IF;")
make_st()
```

### Normalised feature matrix

```python
# Normalise the query features offline, so the PLC only has to compute distances:
# each record stores [SP(i), SP(i-1), RPM(i), RPM(i-1), PWM(i-1), PWM(i-2)] / range.
df = pd.read_csv("output/m340/DD_E_FRIT_Optimized_Dataset.csv")

sp_range = df["SetPoint"].max() - df["SetPoint"].min()
rpm_range = df["RPM"].max() - df["RPM"].min()
pwm_range = df["PWM"].max() - df["PWM"].min()

inv_sp = 1.0 / sp_range if sp_range != 0 else 1.0
inv_rpm = 1.0 / rpm_range if rpm_range != 0 else 1.0
inv_pwm = 1.0 / pwm_range if pwm_range != 0 else 1.0

print("inv_sp =", inv_sp)
print("inv_rpm =", inv_rpm)
print("inv_pwm =", inv_pwm)

features = []
pids = []

for i in range(2, len(df)):
    sp_i = df.loc[i, "SetPoint"]
    sp_prev = df.loc[i - 1, "SetPoint"]

    rpm_i = df.loc[i, "RPM"]
    rpm_prev = df.loc[i - 1, "RPM"]

    pwm_prev = df.loc[i - 1, "PWM"]
    pwm_prev2 = df.loc[i - 2, "PWM"]

    kp = df.loc[i, "Kp"]
    ki = df.loc[i, "Ki"]
    kd = df.loc[i, "Kd"]

    features.append([
        sp_i * inv_sp,
        sp_prev * inv_sp,
        rpm_i * inv_rpm,
        rpm_prev * inv_rpm,
        pwm_prev * inv_pwm,
        pwm_prev2 * inv_pwm
    ])

    pids.append([
        kp,
        ki,
        kd
    ])

feature_df = pd.DataFrame(
    features,
    columns=[
        "SP_now_norm",
        "SP_prev_norm",
        "RPM_now_norm",
        "RPM_prev_norm",
        "PWM_prev_norm",
        "PWM_prev2_norm"
    ]
)

pid_fast_df = pd.DataFrame(
    pids,
    columns=["Kp", "Ki", "Kd"]
)

feature_df.to_csv("output/m340/DD_featureMatrix.csv", index=False)
pid_fast_df.to_csv("output/m340/DD_pidFastMatrix.csv", index=False)

print("Original dataset size:", len(df))
print("Valid feature rows:", len(feature_df))
```

### Structured Text: feature matrix

```python
# Write the normalised feature matrix and its gains as Structured Text initialisation code.
def make_st_feature_matrix():
    feature_df = pd.read_csv("output/m340/DD_featureMatrix.csv")
    pid_fast_df = pd.read_csv("output/m340/DD_pidFastMatrix.csv")

    n = len(feature_df)

    lines = []

    lines.append("IF %S21 THEN")
    lines.append(f"    DATASET_SIZE := {n};")
    lines.append("")
    lines.append("    (* Feature structure: SP_now, SP_prev, RPM_now, RPM_prev, PWM_prev, PWM_prev2 *)")
    lines.append("")

    for i, row in feature_df.iterrows():
        lines.append(f"    featureMatrix[{i},0] := {row['SP_now_norm']:.8f};")
        lines.append(f"    featureMatrix[{i},1] := {row['SP_prev_norm']:.8f};")
        lines.append(f"    featureMatrix[{i},2] := {row['RPM_now_norm']:.8f};")
        lines.append(f"    featureMatrix[{i},3] := {row['RPM_prev_norm']:.8f};")
        lines.append(f"    featureMatrix[{i},4] := {row['PWM_prev_norm']:.8f};")
        lines.append(f"    featureMatrix[{i},5] := {row['PWM_prev2_norm']:.8f};")
        lines.append("")

        lines.append(f"    pidMatrix[{i},0] := {pid_fast_df.loc[i, 'Kp']:.8f};")
        lines.append(f"    pidMatrix[{i},1] := {pid_fast_df.loc[i, 'Ki']:.8f};")
        lines.append(f"    pidMatrix[{i},2] := {pid_fast_df.loc[i, 'Kd']:.8f};")
        lines.append("")

    lines.append("    bDataLoaded := TRUE;")
    lines.append("END_IF;")

    with open("output/m340/data_init_featureMatrix.txt", "w") as f:
        f.write("\n".join(lines))

make_st_feature_matrix()
```

### Performance metrics

```python
def calculate_metrics(data, name="Dataset"):
    error=data['SetPoint']-data['RPM']
    mae=np.mean(np.abs(error))                    
    rmse=np.sqrt(np.mean(error**2))
    iae=np.sum(np.abs(error))

    delta_u = np.diff(data['PWM'])                 
    tv=np.sum(np.abs(delta_u))
    delta_u_sq=np.sum(delta_u**2)     
    
    max_error = np.max(np.abs(error))
    
    return {
        "Dataset": name,
        "MAE (RPM Error)": round(mae, 3),
        "RMSE (RPM Error)": round(rmse, 3),
        "Integral Absolute Error": round(iae, 3),
        "Max Error (RPM)": round(max_error, 3),
        "Control TV (Sum |ΔPWM|)": round(tv, 3),
        "Control Effort (Sum ΔPWM^2)": round(delta_u_sq, 3)
    }

metrics = [
    calculate_metrics(df, "Initial (Static PID)"),
    calculate_metrics(df2, "Final (Optimized PID)"),
]

metrics_df = pd.DataFrame(metrics)

print("Performance Metrics Comparison")
print(metrics_df.to_string(index=False))
```

### Extended metrics and per-setpoint breakdown

```python
def calculate_metrics(data, name="Dataset"):
    error = data['SetPoint'] - data['RPM']
    mae = np.mean(np.abs(error))                    
    rmse = np.sqrt(np.mean(error**2))
    iae = np.sum(np.abs(error))
    mean_error=np.mean(error)

    delta_u = np.diff(data['PWM'])                 
    tv = np.sum(np.abs(delta_u))
    delta_u_sq = np.sum(delta_u**2) 

    error_std=np.std(error)    
    
    max_error = np.max(np.abs(error))
    
    return {
        "Dataset": name,
        "MAE (RPM Error)": round(mae, 3),
        "RMSE (RPM Error)": round(rmse, 3),
        "Integral Absolute Error": round(iae, 3),
        "Average Mean": round(mean_error, 3),
        "Max Error (RPM)": round(max_error, 3),
        "Control TV (Sum |ΔPWM|)": round(tv, 3),
        "Control Effort (Sum ΔPWM^2)": round(delta_u_sq, 3),
        "Std error": round(error_std, 3)
    }


def calculate_per_setpoint(data, name="Dataset"):
    rows = []
    
    for sp in sorted(data['SetPoint'].unique()):
        subset = data[data['SetPoint'] == sp]
        error = subset['SetPoint'] - subset['RPM']
        
        error_std = np.std(error)
        rpm_std = np.std(subset['RPM'])
        mean_err = np.mean(error)
        mae = np.mean(np.abs(error))
        
        rows.append({
            "Dataset": name,
            "Setpoint": int(sp),
            "Mean Error": round(mean_err, 2),
            "MAE": round(mae, 2),
            "Error Std": round(error_std, 2),
            "RPM Std": round(rpm_std, 2),
            "N samples": len(subset)
        })
    
    return rows

metrics = [
    calculate_metrics(df,  "Initial (Static PID)"),
    calculate_metrics(df2, "Final (Optimized PID)"),
]
metrics_df = pd.DataFrame(metrics)

print("Performance Metrics")
print(metrics_df.to_string(index=False))

per_sp_rows = []
per_sp_rows += calculate_per_setpoint(df,  "Initial (Static PID)")
per_sp_rows += calculate_per_setpoint(df2, "Final (Optimized PID)")

per_sp_df = pd.DataFrame(per_sp_rows)

print()
print(per_sp_df.to_string(index=False))
```
